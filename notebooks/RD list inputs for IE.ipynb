```python
import pandas as pd
```

```python
soto = pd.read_excel("../data/Soto_supplemental_table", sheet_name = "Table S2")
soto_RDs = soto[soto["Domain type"] == "RD"]
soto_RDs["Start"] = soto_RDs["Coordinates"].str.split("-").str[0]
soto_RDs["End"] = soto_RDs["Coordinates"].str.split("-").str[1]
soto_RDs  = soto_RDs[["Uniprot ID", "Sequence", "Start", "End"]]
soto_RDs
```

```python
soto_RDs.to_csv("../data/ie_files/domain_list1_Soto.csv")
```

```python
delrosso = pd.read_excel("../data/Stanford_Supplementary Table 2 Domains from Tiles.xlsx", sheet_name = "Repression Domains")
delrosso = delrosso.rename(columns = {"UniProt ID" : "Uniprot ID"})
delrosso = delrosso[["Uniprot ID", "Sequence", "Start", "End"]]
delrosso
```

```python
delrosso.to_csv("../data/ie_files/repression_domain_list_DelRosso.csv")
```
