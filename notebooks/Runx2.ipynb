```python
def return_big_df(folder_path):
    dfs = []
    for path in glob.glob(folder_path):
        dfs.append(pd.read_csv(path, sep = "\t", header = None, index_col = 0).reset_index())
    return pd.concat(dfs)
```

```python
all_TFs = pd.read_csv("../soto_analysis/outputs/all_TFs_table_proteins.txt", sep = "\t")
all_TFs[all_TFs["uniprotID"] == "Q13950"]
    
```

```python
activator_TFs = all_TFs.dropna(subset = "AD_coords")
activator_TFs
```

```python
zoonomia_all_TFs = return_big_df("../soto_analysis/outputs/mutations/cds_zoonomia_all_TF_cds/*")
# Calculate average per TF
zoonomia_all_TFs = zoonomia_all_TFs[[3,8]].groupby(3).mean().reset_index()
zoonomia_all_TFs
```

```python
verteb_all_TFs = return_big_df("../soto_analysis/outputs/mutations/cds_100_verteb_all_TF_cds/*")
# Calculate average per TF
verteb_all_TFs = verteb_all_TFs[[3,8]].groupby(3).mean().reset_index()
verteb_all_TFs
```

```python
all_TFs = pd.read_csv("../soto_analysis/outputs/all_TFs_table_proteins.txt", sep = "\t")
name_map = pd.read_csv("../data/lambert_tfs_gene_names.tsv", sep = "\t")
name_map = name_map.rename(columns = {"From" : "uniprotID", "Gene Names (primary)" : "Gene"})
name_map = name_map.drop(columns = "Entry")
all_TFs_with_gene_name = pd.merge(all_TFs, name_map, on = "uniprotID")
all_TFs_with_gene_name
```

```python
dfs = []
for path in glob.glob("../data/zoonomia_toga_mca_all_TFs/prot_alignment_percent_identities_for_vis/*"):
    df = pd.read_csv(path, index_col = 0)
    df["path"] = path
    dfs.append(df)
percent_identities = pd.concat(dfs)
percent_identities["Gene"] = percent_identities["path"].str.split("/").str[-1]
percent_identities = percent_identities.drop(columns = "path")
percent_identities = percent_identities[["Gene", "percent_identity"]].groupby("Gene").mean().reset_index()
percent_identities_with_uniprotID = pd.merge(percent_identities, all_TFs_with_gene_name[["uniprotID", "Gene"]])
percent_identities_with_uniprotID
```

```python
cds_mtr = []

cds_paths = glob.glob(f"../soto_analysis/outputs/mutations/cds_mtr/*")

for path in cds_paths:
    file_ENST = path.split("/")[-1].split(".bed")[0]
    try:
        df = pd.read_csv(path, sep = "\t", header = None)
        cds_mtr.append(df)
    except pd.errors.EmptyDataError:
        print(file_ENST, " is empty and has been skipped.")

cds_mtr = pd.concat(cds_mtr)
cds_mtr = cds_mtr.rename(columns = {3: "ENST", 8: "MTR"})    
cds_mtr
```

```python
cds_mtr = pd.merge(cds_mtr, all_TFs_with_gene_name[["ENST", "Gene", "uniprotID"]])
cds_mtr
```

```python
fig,axs = plt.subplots(1,3, dpi = 300, figsize = (9,3))

sns.set_context('talk')

## Zoonomia codon alignment
df = percent_identities_with_uniprotID[~percent_identities_with_uniprotID['uniprotID'].isin(activator_TFs["uniprotID"])]
sns.kdeplot(df["percent_identity"], label = "Non-Activator", fill = True, ax = axs[0])

df = percent_identities_with_uniprotID[percent_identities_with_uniprotID['uniprotID'].isin(activator_TFs["uniprotID"])]
sns.kdeplot(df["percent_identity"], label = "Activator",   fill = True, ax = axs[0])

axs[0].axvline(percent_identities_with_uniprotID[percent_identities_with_uniprotID['uniprotID'] == "Q13950"]["percent_identity"].iloc[0], color = 'red')
axs[0].set_xlabel("Zoonomia\nCodon Alignment\nPercent Identity")

## Zoonomia
zoonomia_all_TFs = zoonomia_all_TFs[[3, 8]].groupby(3).mean().reset_index()

df = zoonomia_all_TFs[~zoonomia_all_TFs[3].isin(activator_TFs["ENST"])]
sns.kdeplot(df[8], label = "Non-Activator",  fill = True, ax = axs[1])

df = zoonomia_all_TFs[zoonomia_all_TFs[3].isin(activator_TFs["ENST"])]
sns.kdeplot(df[8], label = "Activator",  fill = True, ax = axs[1])

axs[1].axvline(df[df[3] == "ENST00000647337"][8].iloc[0], color = 'red')
axs[1].set_xlabel("Zoonomia\nPhyloP")

## 100 Vertebrates
verteb_all_TFs = verteb_all_TFs[[3, 8]].groupby(3).mean().reset_index()

df = verteb_all_TFs[~verteb_all_TFs[3].isin(activator_TFs["ENST"])]
sns.kdeplot(df[8], label = "Non-Activator",  fill = True, ax = axs[2])

df = verteb_all_TFs[verteb_all_TFs[3].isin(activator_TFs["ENST"])]
sns.kdeplot(df[8], label = "Activator",  fill = True, ax = axs[2])

axs[0].annotate('', xy=(0.05, 0.95), xycoords='axes fraction', 
            xytext=(0.35, 0.95), textcoords='axes fraction',
            arrowprops=dict(arrowstyle='<|-', color='gray', lw=2))

axs[0].text(0.4, 1.1, "Conservation", transform=axs[0].transAxes, fontsize='x-small', va='top', ha='center', color = 'gray')

axs[2].axvline(df[df[3] == "ENST00000647337"][8].iloc[0], color = 'red', label = "RUNX2")
axs[2].set_xlabel(" 100 Vertebrates\nPhyloP")

axs[1].set_ylabel("")
axs[2].set_ylabel("")

axs[2].legend(bbox_to_anchor = (1,1), title = "TF Type")

plt.tight_layout(pad = 0.25)

fig.supxlabel("Average per TF", y = -0.1, x= 0.45)
sns.despine()
```
