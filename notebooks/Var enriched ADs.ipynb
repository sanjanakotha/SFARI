```python
import pandas as pd
```

```python
SFARI_ADs = pd.read_csv("../output/SFARI_ADs_considering_isoforms_and_canonical.csv", index_col = 0)
SFARI_ADs
```

```python
sibling_result = pd.read_csv("../output/sibling_analysis_results_fixed.csv")
sibling_result
```

```python
proband_only_hits = sibling_result[sibling_result["proband_only AD_vs_DBD_sig"] 
| sibling_result["proband_only AD_vs_TF_sig"]
| sibling_result["proband_only AD_vs_IDR_sig"]]
proband_only_hits
```

```python
for col in sibling_result.columns:
    if col.startswith("proband_only") and sibling_result[col].dtype == bool:
        print(f"Gene symbols where {col} is True:")
        print(sibling_result[sibling_result[col]]["gene-symbol"].tolist())
        print()
```

```python
hit_ADs = SFARI_ADs[SFARI_ADs["Gene"].isin(proband_only_hits["gene-symbol"])]
hit_ADs = hit_ADs[["Gene", "uniprotID", "Start", "End", "ProteinRegionSeq", "Reference"]]
hit_ADs
```

```python
hit_ADs.to_csv("../output/proband_only_hit_iwes_v3_SFARI_ADs.csv")
```
