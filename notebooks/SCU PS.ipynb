```python
import pandas as pd
```

```python
PS_plus = pd.read_csv("../data/SCU_phosphoserines.tsv", sep = "\t")
PS_plus["Modified residue_split"] = PS_plus["Modified residue"].str.split("; MOD_RES")
PS_plus
```

```python
PS_plus_split = PS_plus.explode("Modified residue_split")
PS_plus_split = PS_plus_split.dropna(subset = "Modified residue_split")
PS_plus_split = PS_plus_split[PS_plus_split["Modified residue_split"].str.contains("Phosphoserine")]
PS_plus_split["pos"] = PS_plus_split["Modified residue_split"].str.extract(r'(\d+);').astype(int)
PS_plus_relev_cols = PS_plus_split[["From", "Entry", "pos"]]
PS_plus_relev_cols
```

```python
known_ADs = pd.read_csv("../output/known_ADs_considering_isoforms_and_canonical.csv")
known_ADs
```

```python
PS_plus_relev_cols_ad_pos = pd.merge(PS_plus_relev_cols, known_ADs[["Gene", "Start", "End"]], left_on = "From", right_on = "Gene", how = 'left')
PS_plus_relev_cols_ad_pos
```
