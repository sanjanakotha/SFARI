```python
def plot_comparison(output, ax=None, domains_to_plot=None, legend=True):
    """
    Plot missense SNP frequencies for selected domains using Seaborn colorblind palette.
    
    Parameters:
    - output: DataFrame with columns like 'AD_missense_prop', 'DBD_missense_prop', etc.
    - ax: matplotlib axes object. If None, a new figure and axes are created.
    - domains_to_plot: list of domain names to plot. Options: 'AD', 'DBD', 'IDR_outside_AD', 'TF_outside_AD'.
                       Default: all of them.
    """
    sns.set_context('poster')
    palette = sns.color_palette("colorblind")

    # Fixed mapping of domain to palette index
    domain_colors = {
        "TF_outside_AD": palette[0],
        "AD": palette[1],
        "DBD": palette[2],
        "IDR_outside_AD": palette[3]
    }

    if ax is None:
        fig, ax = plt.subplots(figsize=(10, 6), dpi=300)

    # Compute TF outside AD if needed
    rest_of_TF_prop = (output["TF_missense"] - output["AD_missense"]) \
                      / (output["TF_cds_length"] - output["AD_cds_length"])

    # Default domains
    if domains_to_plot is None:
        domains_to_plot = ["TF_outside_AD", "AD", "DBD", "IDR_outside_AD"]

    for domain in domains_to_plot:
        color = domain_colors.get(domain)
        if domain == "TF_outside_AD":
            sns.kdeplot(rest_of_TF_prop, label="TF outside AD", color=color, ax=ax)
        elif domain == "AD":
            sns.kdeplot(output["AD_missense_prop"], label="AD", color=color, ax=ax)
        elif domain == "DBD":
            sns.kdeplot(output["DBD_missense_prop"], label="DBD", color=color, ax=ax)
        elif domain == "IDR_outside_AD":
            sns.kdeplot(output["IDR_missense_prop"], label="IDR outside AD", color=color, ax=ax)
        else:
            raise ValueError(f"Unknown domain: {domain}")

    ax.set_xlabel("Missense SNP Frequency")
    if legend:
        ax.legend(bbox_to_anchor = (1,1))
    sns.despine(ax=ax)
    
    return ax

```

```python
fig, axs = plt.subplots(2,2,sharey='row', sharex='row', width_ratios=(1, 1), figsize = (10, 6), dpi = 300)
axs = axs.flatten()


plot_comparison(all_proband_output, ax = axs[0], legend=False)
plot_comparison(all_sibling_output, ax = axs[1], legend=False)
plot_comparison(proband_only_output, ax = axs[2], legend=False)
plot_comparison(sibling_only_output, ax = axs[3], legend=False)

# Add descriptive labels to each subplot
set_labels = ["All\nProband", "All\nSibling", "Proband\nOnly", "Sibling\nOnly"]
for ax, label in zip(axs, set_labels):
    # Position in axes coordinates (0,1 is top-left)
    ax.text(0.02, 0.95, label, transform=ax.transAxes, fontsize=12,
            verticalalignment='top', ha = 'left')
    
for ax in axs:
    ax.set_xlabel("")
    ax.set_ylabel("")

# Add a single legend outside the plot
handles, labels = axs[0].get_legend_handles_labels()  # get legend from first plot
fig.legend(handles, labels, bbox_to_anchor=(0.95,0.95), fontsize=12, title="Domain", title_fontsize=12)

fig.tight_layout(rect=[0, 0, 0.85, 1], pad =0.5)  # leave space on the right for legend
fig.supxlabel("Missense SNP Proportion", y = -0.05, x = 0.45)
fig.supylabel("Density", x = -0.05)
# fig.tight_layout(pad=0.5)
```
